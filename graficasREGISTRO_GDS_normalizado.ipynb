```python
import random
import matplotlib.pyplot as plt

datos = [
(4.4,6,90.2,191,15.16),
(3.6,6.6,66.2,10,11.6),
(1.7,7.7,75.9,150,12.88),
(6.7,8,52.9,170,15.75),
(10.2,4.6,65.6,135,17.27),
(3.7,9.4,46.4,20,11.45),
(2.9,7.9,81,177,14.95),
(0.9,4.2,75,178,11.48),
(8.1,5.8,45.5,14,12.22),
(2.3,4.6,61.6,2,8.84),
(2.9,4.9,44,76,9.73),
(10.9,6.3,82,64,20),
(1,4.1,92.1,52,10.16),
(10.1,4.6,44.4,120,15.63),
(9,8.7,80.2,36,17.23),
(4.7,6.8,55.8,103,13.07),
(11.9,7.4,60.9,154,20),
(6.2,5.8,77.3,4,11.86),
(7.9,9.2,69.8,176,18.28),
(7.4,6.8,73.5,200,16.73),
(1.8,4.7,93,10,10.68),
(9.4,6.5,62.3,101,17.55),
(8.6,4,82.7,100,17.26),
(11.2,4.9,94.3,143,20),
(9.2,8.6,51.6,198,18.86),
(7.6,9.4,76.9,175,19.08),
(2.8,8.6,70.5,47,12),
(7.6,4.6,68.4,70,15.39),
(0.8,6,92.5,120,13.33),
(2.4,7.9,93.9,167,14.99),
(3.2,7.4,43.6,23,8.97),
(5.5,9.5,90.2,142,17.08),
(3.1,5,96.1,102,12.98),
(1.2,4.9,64.8,73,8.64),
(2.8,5.9,91.9,83,12.6),
(5.7,4.8,80.3,97,14.56),
(8.1,9.8,53.9,65,17.35),
(6,5.4,43,140,14.05),
(5.4,7.9,45.9,41,13.34),
(1.9,9.9,46,44,10.32),
(8.9,7.5,40.3,175,17.76),
(4.2,6.2,78.4,144,14.54),
(7.9,4.5,56.7,30,12.29),
(1.3,7.7,72.9,136,13.03),
(4.4,6.8,78.7,138,14.02),
(5.6,7.9,97.5,1,14.87),
(11.9,4.6,48.5,194,19.67),
(0.6,4.7,95.1,36,10.19),
(4.7,9.6,50,94,12.61),
(6.4,5.1,75.7,140,14.93),
(11.5,5.3,89.6,178,20),
(11.7,6.8,63.1,78,17.74),
(0.9,8.2,62.5,138,10.33),
(10.6,4.6,70.3,89,17.5),
(9,6.3,46.1,73,14.4),
(7.6,5.2,42.9,115,14.31),
(11,4.6,98.5,184,20),
(6.5,9.1,71,41,14.44),
(10,5.7,48.7,178,17.45),
(0.1,4.4,79.8,181,12.79),
(5,7.4,40.5,101,12.67),
(2,10,65.5,166,13.11),
(1.4,9.2,89.8,76,14.42),
(2.3,8.8,57.7,135,12.29),
(8.6,7.5,43.5,185,16.63),
(5.2,6.9,85.9,30,15.24),
(5,9.6,87.3,29,14.55),
(8.5,6.9,80.6,197,18.72),
(6.9,9.8,45.9,166,15.84),
(1.2,8.1,76.1,109,12.33),
(10.6,6.5,75.3,34,17.71),
(4.5,6,52.8,172,13.64),
(5.8,8.1,58,19,12.93),
(0.6,9.9,64.5,9,10.25),
(3.6,8.7,45.5,80,11.81),
(10.9,7.7,44.4,148,18.2),
(11.2,4.3,56.4,112,17.7),
(8.2,6.6,43.9,110,16.44),
(7.2,9.1,89.5,132,18.32),
(4.8,5.8,91.2,109,16.13),
(2.8,6.4,49,165,11.55),
(3.1,9.3,83.3,74,13.85),
(9.2,4.8,78.1,142,18.02),
(9.8,7.3,54,168,18.39),
(9.4,8.4,87.6,127,19.44),
(8.7,8.4,96.2,8,17.59),
(10.4,6,45.7,133,17.96),
(7.8,9.5,92.8,1,16.56),
(2.5,4.2,68.7,0,9.49),
(6.9,5.8,63.7,31,13.78),
(7.7,7.6,52.7,167,17.07),
(10.4,4.2,63,46,17.04),
(6.6,7.2,77.1,188,16.96),
(10.9,4.5,69.9,162,17.91),
(1.8,9,79.8,38,12.99),
(11.1,9.3,59.4,104,19.24),
(11,6.1,90.6,182,20),
(10.1,6.4,71.9,125,18.55),
(5.4,4.3,79.8,139,14.98),
(9.8,7.4,87.6,104,18.36),
(2.5,8.4,40.6,15,9.36),
(6.4,5.7,98.1,141,17.32),
(1.6,8.1,69.5,194,12.98),
(10.6,4.3,71.8,31,17.14),
(0.7,4.4,75.6,88,10.23),
(7.5,4,65.7,25,14.61),
(7.9,5.5,48.6,62,14.87),
(7.7,8.5,72.2,200,19.4),
(5.2,8.1,86.6,20,14.48),
(5.5,6,70.6,46,13.35),
(1.4,6.4,44.2,150,10.38),
(7.9,9.4,49.3,60,14.2),
(0.5,6.8,45.9,139,9.56),
(4.9,8.2,95.9,178,16.82),
(7.5,4.7,44.5,119,15.31),
(4.6,5.6,66.6,87,12.78),
(5.6,6.6,57.2,2,12.28),
(4.8,7.5,52.4,158,14.29),
(10.3,7,74.7,5,16.91),
(8.8,8.4,64.4,168,18.49),
(8.3,9.3,93.1,110,18.41),
(9.8,5.4,97.5,112,19.9),
(6.4,7.2,87.3,19,13.93),
(10.3,7.2,44.1,100,18.87),
(10.4,8.7,59,19,17.08),
(8.7,4.7,64.2,52,15.12),
(5.8,9.9,60.4,29,14.47),
(4.8,9,74.4,128,15.21),
(4.2,9.5,61,177,14.8),
(6.1,5,62.4,157,14.83),
(3.9,4,48.4,116,10.42),
(9.4,9.9,69.3,151,20),
(11.1,9.5,73.9,147,20),
(6.5,8.4,95.5,152,19.48),
(10.8,8.2,68.3,16,17.74),
(5,5.2,48.8,30,13.2),
(8.5,4.8,57.4,94,13.45),
(10.8,6.1,65.9,56,17.91),
(8.8,5.5,56.6,198,18.54),
(7.1,7.6,61.5,23,14.74),
(3,7.7,57.3,68,11.05),
(11.2,7.6,77.3,127,20),
(11.7,9.3,59.7,11,18.33),
(1.3,5.6,84.9,65,10.78),
(6.4,9,81.4,129,17.96),
(4.5,7.4,76.6,75,12.24),
(10.7,6.8,94.4,153,19.61),
(10.7,8.7,78.2,9,18.31),
(3.2,7.5,44.3,100,11.53),
(10.1,8.1,62.2,71,18.57),
(4,4.9,42.5,158,12.84),
(11,6,66.6,54,19.33),
(9.3,5.4,84.5,195,19.05),
(2.9,4.2,94.5,38,12.56),
(4.1,9.6,68.1,192,16.31),
(6.3,8.2,95.2,54,15.95),
(2.8,6.9,60.8,61,11.51),
(2.5,8.6,49.3,170,12.63),
(5.4,8.4,44.9,113,12.86),
(4.2,7.1,43,66,11.74),
(4.8,7.8,93.8,77,15.66),
(3.9,9.4,95.5,100,16.35),
(2.2,9.5,96.8,151,14.62),
(6.3,8.9,50.8,8,13.57),
(1.5,7.1,44.8,93,10.54),
(7.2,8.9,84.3,27,16.15),
(11.2,5.3,58.8,20,16.5),
(2.6,4.6,95.3,22,12.15),
(11.5,7,47.9,138,19.15),
(1.6,8.6,50.9,45,10.26),
(8.1,9.8,82.3,27,16.74),
(9.2,5.4,90.9,183,20),
(2.5,5.3,42.8,58,9.28),
(11.5,4.3,51.5,60,16.93),
(9.5,9.2,74.1,54,19.48),
(8.5,9.3,99.5,109,19.43),
(6.7,8.9,66,103,17.29),
(11.7,6.7,62.3,24,17.38),
(7.4,9.1,99.9,129,19.38),
(1.5,4.3,93.1,48,9.59),
(11.4,7.2,42.6,90,17.19),
(1.4,8.5,92.1,181,14.39),
(2.3,7.2,90.8,91,12.46),
(1.4,5.1,87,177,13.13),
(0.2,6.2,47.2,177,10.32),
(3.8,8.8,44,68,11.58),
(10,4.9,82.2,30,18.19),
(4.2,9.3,45.6,36,10.2),
(5.2,5,80.1,131,14.99),
(1.4,8.1,71.9,62,11.45),
(3.6,7.4,95,118,14.24),
(0.5,9.1,97.3,149,13.72),
(7.2,4.5,78.9,100,15.72),
(5.4,6.6,88.2,44,14.36),
(11.3,6.1,50.5,177,19.09),
(2.9,4.4,82.6,158,13.67),
(5.1,5.4,82.3,87,14.36),
(9.3,9.2,42.2,142,16.55),
(3.7,6,48.7,24,10.55),
(11.6,4.6,90.6,169,20)
]

# ============================================================
# FUNCIONES
# ============================================================

def normalizar(x):
    xmin = min(x)
    xmax = max(x)
    return [(xi - xmin)/(xmax - xmin) for xi in x], xmin, xmax

def predecir(x,w,b):
    return w*x + b

def mse(X,Y,w,b):
    return sum((y - predecir(x,w,b))**2 for x,y in zip(X,Y))/len(X)

def mae(X,Y,w,b):
    return sum(abs(y - predecir(x,w,b)) for x,y in zip(X,Y))/len(X)

def gradiente(x,y,w,b):
    error = y - (w*x + b)
    return -2*x*error, -2*error

def sgd(X,Y,lr=0.05,epocas=10):
    w,b = 0,0
    random.seed(1)

    print("\n============================================================")
    print("  STOCHASTIC GRADIENT DESCENT (SGD)")
    print("============================================================")
    print(f"  lr={lr}  |  N={len(X)}  |  epocas={epocas}")
    print(f"{'Epoca':>6} {'w':>10} {'b':>10} {'MSE':>12}")
    print("-"*50)

    for e in range(epocas):
        indices = list(range(len(X)))
        random.shuffle(indices)

        for i in indices:
            gw,gb = gradiente(X[i],Y[i],w,b)
            w -= lr*gw
            b -= lr*gb

        error = mse(X,Y,w,b)

        nota = ""
        if e==0: nota="<-- inicio"
        if e==epocas-1: nota="<-- fin"

        print(f"{e:6} {w:10.4f} {b:10.4f} {error:12.4f} {nota}")

    print(f"\nResultado: w={w:.4f}, b={b:.4f}")
    print(f"MSE={mse(X,Y,w,b):.4f}")
    print(f"MAE={mae(X,Y,w,b):.4f}")

    return w,b

def graficar(X,Y,w,b,titulo):
    plt.figure()
    plt.scatter(X,Y)
    x_line = [min(X),max(X)]
    y_line = [predecir(x,w,b) for x in x_line]
    plt.plot(x_line,y_line)
    plt.title(titulo)
    plt.xlabel("Variable (normalizada)")
    plt.ylabel("Nota final")
    plt.show()

horas_estudio = [d[0] for d in datos]
horas_suenio = [d[1] for d in datos]
asistencia = [d[2] for d in datos]
ejercicios = [d[3] for d in datos]
nota = [d[4] for d in datos]

variables = [
("Horas Estudio", horas_estudio),
("Horas Sueño", horas_suenio),
("Asistencia %", asistencia),
("Ejercicios", ejercicios)
]

for nombre, X in variables:
    print("\n"+"#"*60)
    print(f" VARIABLE: {nombre}")
    print("#"*60)

    X_norm,_,_ = normalizar(X)
    Y_norm,_,_ = normalizar(nota)

    w,b = sgd(X_norm,Y_norm)

    graficar(X_norm,Y_norm,w,b,f"{nombre} vs Nota Final")
```
